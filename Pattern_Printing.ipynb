# Patten Printing Programs

```python
#Q1
n = 4

for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

```python
#Q2
n = int(input("Enter No of rows: "))

for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

```python
#Q3
rows = int(input("Enter No of rows: "))
columns = int(input("Enter No of columns: "))

for i in range(rows):
    for j in range(columns):
        print("*", end=" ")
    print()
```

```python
#Q4
n = 5

for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print(end="\n")
```

```python
#Q5
n = 5

for i in range(n, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print(end="\n")
```

```python
#Q6
n = int(input("Enter no of rows: "))

for i in range(1, n + 1):
    for j in range(0, n - i):
        print(" ", end="")
    for j in range(0, i):
        print("* ", end="")
    print(end="\n")
```

```python
#Q7
n = 5

for i in range(n, 0, -1):
    for j in range(0, n - i):
        print(" ", end="")
    for j in range(0, i):
        print("* ", end="")
    print(end="\n")
```

```python
#Q8
n = 5

for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("* ", end="")
    print(end="\n")
```

```python
#Q9
row = 5
num = 2

for i in range(1, row + 1):
    for j in range(1, i + 1):
        print(num, end=" ")
        num += 1
    print(end="\n")
```

```python
#Q10
n = 5

for i in range(1, n + 1):
    for j in range(0, i):
        print(chr(65 + j), end=" ")
    print(end="\n")
```

```python
#Q11
n = 4
row = 1

while row <= n:
    star = 1
    while star <= n:
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
#Q12
num = 5
row = 1

while row <= num:
    star = 1
    while star <= row:
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
#Q13
n = 5

for i in range(1, 2 * n):
    if i <= n:
        num = i
    else:
        num = 2 * n - i

    for j in range(1, num + 1):
        print(j, end=" ")

    print(end="\n")
```

```python
#Q14
n = 5

for i in range(1, 2 * n):
    if i <= n:
        star = i
    else:
        star = 2 * n - i

    space = n - star

    for j in range(space):
        print(" ", end=" ")

    for k in range(2 * star - 1):
        print("*", end=" ")

    print("\n")
```

```python
#Q15
num = 5
row = 1

while row <= num:
    space = 1
    while space <= num - row:
        print(" ", end=" ")
        space += 1

    star = 1
    while star <= 2 * row - 1:
        print("*", end=" ")
        star += 1

    print(end="\n")
    row += 1
```

```python
#Q16
num = 5
row = 1

while row <= num:
    space = 1
    while space <= row - 1:
        print(" ", end=" ")
        space += 1

    star = 1
    while star <= num - row + 1:
        print("*", end=" ")
        star += 1

    print(end="\n")
    row += 1
```
