```python
#Q1
class student:
    def __init__(self, Name, Roll_No, Marks):
        self.Name = Name
        self.Roll_No = Roll_No
        self.Marks = Marks

    def calculate_total(self):
        return sum(self.Marks)

    def calculate_average(self):
        return self.calculate_total()/len(self.Marks)

    def calculate_grade(self):
        average = self.calculate_average()

        if average >=90:
            return "A+"

        elif average>=80:
            return "A"

        elif average >=70:
            return "B"

        elif average >= 60:
            return "C"

        elif average >= 40:
            return "D"

        else:
            return "F"

    def display_report(self):
        print("======Student Report=======")
        print("Name of Student : ",self.Name)
        print("ROll no. : ", self.Roll_No)
        print("Marks : ", self.Marks)
        print("Total Marks : ",self.calculate_total())
        print("Average Marks : ",self.calculate_average())
        print("Grade : ", self.calculate_grade())

#create two student objects
student1=student("Ankit",101,[89,76,98,76,88])
student2=student("Rahul",102,[96,87,69,85,79])

#display data of students
student1.display_report()
student2.display_report()

        
        
    
```

```python
#Q2
class Bank_Account:
    def __init__(self,Account_Holder,Opening_Balance=0):
        if Opening_Balance < 0:
            raise ("Opening Balance Cannot Be Nagative.")
        
        self.Acc_Holder_Name= Account_Holder
        self.Balance= Opening_Balance

    def deposit(self,Amount):
        if Amount <=0:
            return "Deposit amount must be positive"

        else:
            self.Balance +=Amount
            print("Deposit Successfully.")
            print("Deposited Amount :",Amount)

    def withdraw(self,Amount):
        if Amount<=0:
            print("Withdrawal Amount should be greater then Zero.")
        elif Amount < self.Balance:
            print("Insufficient Balnce.")

        else:
            Withdrawal_Amount=self.Balance-Amount
            print("Withdrawal Successful.")
            print("Withdrawn Amount : ", Withdrawal_Amount)


    def check_Balance(self):
        print("Account_Holder : ",self.Acc_Holder_Name)
        print("Current Balance : ",self.Balance)

#Create Objects of Bank Account
Account1=Bank_Account("Ansh",3000)
Account2=Bank_Account("Yash",5000)

#Display Details
Account1.check_Balance()
Account1.deposit(2500)
Account1.withdraw(600)
Account1.check_Balance()

print("\nSecond Account:")
Account2.check_Balance()

            
        
```

```python
#Q3
class Employee:
    def __init__(self,Emp_Id,Name,Basic_Salary):
        self.Employee_ID=Emp_Id
        self.Employee_Name=Name
        self.Basic_Salary=Basic_Salary

    def calculate_HRA(self):
        return self.Basic_Salary*0.20
        
    def calculate_DA(self):
        return self.Basic_Salary*0.10

    def calculate_PF(self):
        return self.Basic_Salary*0.12

    def calculate_Net_Salary(self):
        return (self.Basic_Salary+self.calculate_HRA()+self.calculate_DA()-self.calculate_PF())

    def display_Salary_Slip(self):

        print("==========SAlARY SLIP============")
        print("Employee Id : ",self.Employee_ID )
        print("Employee Name : ",self.Employee_Name)
        print("Employee Basic Salary : ",self.Basic_Salary)
        print("Employee HRA(20%) : ",self.calculate_HRA())
        print("Employee DA(10%) : ",self.calculate_DA())
        print("Employee PF(12%) : ",self.calculate_PF())
        print("Employee Net Salary : ",self.calculate_Net_Salary())

#create employee objects
Employee01=Employee(101,"Rahul",20000)
Employee02=Employee(102,"Saket",25000)

#display Salary Details

Employee01.display_Salary_Slip()
Employee02.display_Salary_Slip()




```
