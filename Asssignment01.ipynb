```python
#without list comp...
data=[1,2,3,4,5]
new_Data=[]

for i in data:
    new_Data.append(i**2)

print(data)
print(new_Data)
```

```python
#with list comp...
data=[1,2,3,4,5]
new_Data=[i**2 for i in data]
print(new_Data)
```

```python
data=[23,43,56,34,587,435,346]
new_data=[ex for ex in data if ex%2==0]
print(data)
print(new_data)
```

```python
marks=[98,43,65,87,54,34,95]
result=["Pass" if ex>=60 else "fail" for ex in marks]
print(result)
```

```python
Names=["Ansh","Rohit","Kunal","Gautam"]
result=[ex.upper() for ex in Names]
print(result)
```

```python
Names=["Ansh  ","   Rohit","   Kunal  ","  Gautam  "]
result=[ex.strip() for ex in Names]
print(result)
```
